```python
from google.colab import drive
drive.mount("/content/drive")

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

gpus = tf.config.list_physical_devices('GPU')

if gpus:
    print(f"GPU detectada com sucesso: {gpus[0].name}")
else:
    print("Nenhuma GPU encontrada. Verifique se ativou a GPU nas configurações do ambiente.")

from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import (Conv2D,MaxPooling2D,Flatten,Dense,Dropout,Rescaling)

DIR = "/content/drive/MyDrive/Assistente de Inteligência Artificial - FATESG/Aula - 09/dataset_raiox_reduzido/"
caminhoTreino = DIR + "treino"
caminhoTeste = DIR + "teste"
caminhoProducao = DIR + "producao"

dados_treino = tf.keras.utils.image_dataset_from_directory(caminhoTreino,image_size=(224, 224),batch_size=32,label_mode="binary",shuffle=True)
dados_teste = tf.keras.utils.image_dataset_from_directory(caminhoTeste,image_size=(224, 224),batch_size=32,label_mode="binary",shuffle=False)

print("\nClasses identificadas:")
print(dados_treino.class_names)

modelo = Sequential([Rescaling(1./255,input_shape=(224, 224, 3)),

Conv2D(32,(3, 3),activation="relu"),
MaxPooling2D((2, 2)),Conv2D(64,(3, 3),activation="relu"), #diminui a imagem pela metade (112x112) - A convolução exige menos pixels e o processamento é mais rápido
MaxPooling2D((2, 2)),Dropout(0.25),Conv2D(128,(3, 3),activation="relu"), #diminui a imagem pela metade (56x56)
MaxPooling2D((2, 2)), #diminui a imagem pela metade (28x28)
Dropout(0.25),
Conv2D(256,(3, 3),activation="relu"),
MaxPooling2D((2, 2)), #diminui a imagem pela metade (14x14)- É preciso fazer conta para saber o limite da diminuição sem impactar a análise
Dropout(0.25), #desliga os neurônios temporariamente

Flatten(), # vai empilhar as imagens no formato de 14x14 pixels

Dense(512,activation="relu"),
Dropout(0.50),
Dense(1,activation="sigmoid")
])

modelo.summary()

earlyStopping = EarlyStopping(monitor="val_loss",patience=8,restore_best_weights=True,verbose=1)

modelo.compile(optimizer="adam",loss="binary_crossentropy", metrics=["accuracy"])

historicoAprendizado = modelo.fit(dados_treino,epochs=40,validation_data=dados_teste, callbacks=[earlyStopping])

pd.DataFrame(historicoAprendizado.history).plot(figsize=(10, 6))

plt.grid(True)
plt.xlabel("Época")
plt.ylabel("Valor")
plt.title("Histórico de Aprendizado da CNN")
plt.show()

perda, acuracia = modelo.evaluate(dados_teste)

print("\nResultados:")
print(f"Loss: {perda:.4f}")
print(f"Acurácia: {acuracia:.4f}")
print(f"Acurácia: {acuracia * 100:.2f}%")

caminhoImagem = (caminhoProducao +"/BACTERIA-7422-0001.jpeg")
imagemTeste = load_img(caminhoImagem,target_size=(224, 224))
imagemArray = img_to_array(imagemTeste)
imagemArray = np.expand_dims(imagemArray,axis=0)

previsao = modelo.predict(imagemArray)
probabilidade = previsao[0][0]

print(f"\nProbabilidade retornada pela rede: "f"{probabilidade:.4f}")

if probabilidade > 0.5:
    print("A imagem foi classificada como 'doente'.")
else:
    print("A imagem foi classificada como 'normal'.")
```
